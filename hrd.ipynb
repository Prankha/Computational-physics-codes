```python

import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

from astroquery.mast import Observations
from astroquery.mast import Catalogs
import pandas as pd
import astropy.io.fits as fits
```

```python
# Querying SDSS Legacy Imaging for M13
obs_table = Observations.query_criteria(
    objectname="M44", provenance_name="SDSS Legacy Imaging"
)

# Displaying the table
obs_table
```

```python
# Viewing available products
products = Observations.get_product_list(obs_table)
products
```

```python
# Selecting .jpg image
img_products = Observations.filter_products(
    products,
    # Select .jpg file for photoObj
    productFilename="frame-irg-005045-2-0124.jpg",
)
```

```python
# Downloading preview images
Observations.download_products(img_products, flat=True, verbose=False)
```

```python
img = mpimg.imread("frame-irg-005045-2-0124.jpg")
plt.figure(figsize=(10, 8))
imgplot = plt.imshow(img)
plt.axis("off")
plt.title("Preview Image of M13")
plt.show()
```

```python
# Masking the data to extract only the full FITS file
new_products = Observations.filter_products(
    products,
    # Specify FITS files only
    productGroupDescription="Minimum Recommended Products",
    # Select photoObj files (ugriz)
    productSubGroupDescription="SDSS Imaging Catalogs",
    # Select photoObj-003226-5-0126.fits
    productFilename="photoObj-005045-2-0124.fits",
)

# Downloading photoObj-003226-5-0126.fits
manifest = Observations.download_products(new_products, verbose=False)
```

```python
# Opening file
star_array = fits.open(manifest["Local Path"][0])

# Display header info
star_array.info()
```

```python
# Sending the PSFMAG data to a numpy array
magnitudes = star_array[1].data["PSFMAG"]
magnitudes = np.array(magnitudes)


# Creating arrays for g-band and r-band magnitudes
b_band = magnitudes[:, 1]
v_band = magnitudes[:, 2]

# Calculating color (g - r)
color_data1 = b_band - v_band
```

```python
fig, ax = plt.subplots(figsize=(8, 6), facecolor="white")

# White plot background
ax.set_facecolor("white")

# Scatter points
ax.scatter(
    color_data1,
    v_band,
    s=3,
    color="black",
    label="M44 data"
)

# Limits
ax.set_ylim(10, 30)
ax.invert_yaxis()
ax.set_xlim(-2, 2)

# Labels and title
ax.set_title("CMD of M44", color="black")
ax.set_ylabel("Apparent Magnitude (r-band)", color="black")
ax.set_xlabel("Color (g - r)", color="black")

# Force axes, ticks and tick labels to black
ax.tick_params(
    axis="both",
    colors="black",
    labelcolor="black"
)

for spine in ax.spines.values():
    spine.set_color("black")

# Grid
ax.grid(True, color="gray", alpha=0.3)

ax.legend()

plt.tight_layout()

plt.savefig(
    "CMD_of_M44.png",
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as anim
import random as rand


# ============================================================
# Vector class
# ============================================================

class vec:
    def __init__(self, x: float, y: float):
        self.x = x
        self.y = y

    def __add__(self, other):
        return vec(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return vec(self.x - other.x, self.y - other.y)

    def __mul__(self, scalar: float):
        return vec(self.x * scalar, self.y * scalar)

    def __rmul__(self, scalar: float):
        return self.__mul__(scalar)

    def __truediv__(self, scalar: float):
        return vec(self.x / scalar, self.y / scalar)

    def __iadd__(self, other):
        self.x += other.x
        self.y += other.y
        return self

    def magnitude(self):
        return np.sqrt(self.x**2 + self.y**2)

    def __repr__(self):
        return f"vec({self.x}, {self.y})"

    def gx(self):
        return self.x

    def gy(self):
        return self.y


# ============================================================
# Particle class
# ============================================================

class p:
    def __init__(
        self,
        q: float,
        vx: float,
        vy: float,
        x: float,
        y: float,
        mass: float
    ):
        self.q = q
        self.vel = vec(vx, vy)
        self.mass = mass
        self.pos = vec(x, y)

    def __repr__(self):
        return (
            f"p(q={self.q}, "
            f"pos={self.pos}, "
            f"vel={self.vel}, "
            f"mass={self.mass})"
        )


# ============================================================
# Draw walls
# ============================================================

def plot_walls(ax, width, height):

    # Bottom
    ax.plot([0, width], [0, 0], "k-")

    # Top
    ax.plot([0, width], [height, height], "k-")

    # Left
    ax.plot([0, 0], [0, height], "k-")

    # Right
    ax.plot([width, width], [0, height], "k-")


# ============================================================
# Simulation parameters
# ============================================================

n = 2

T = 100

lx = 10
ly = 11

dt = 0.01


# ============================================================
# Create particles
# ============================================================

particles = []

for i in range(n):

    vx = rand.uniform(0, 5)
    vy = rand.uniform(0, 5)

    x = rand.uniform(0, lx)
    y = rand.uniform(0, ly)

    particle = p(
        q=1,
        vx=vx,
        vy=vy,
        x=x,
        y=y,
        mass=1
    )

    particles.append(particle)


# Print initial particles
for particle in particles:
    print(particle)


# ============================================================
# Figure
# ============================================================

fig, ax = plt.subplots()

ax.set_xlim(0, lx)
ax.set_ylim(0, ly)

ax.set_aspect("equal")

plot_walls(ax, lx, ly)


# ============================================================
# Create graphical objects for particles
# ============================================================

particle_plots = []

for particle in particles:

    particle_plot, = ax.plot(
        [particle.pos.x],
        [particle.pos.y],
        "ro"
    )

    particle_plots.append(particle_plot)


# ============================================================
# Trajectories
# ============================================================

trajectory_plots = []

for particle in particles:

    trajectory_plot, = ax.plot(
        [],
        [],
        "b-",
        linewidth=1
    )

    trajectory_plots.append(trajectory_plot)


# Store trajectory coordinates
x_history = [[] for _ in range(n)]
y_history = [[] for _ in range(n)]


# ============================================================
# Animation update function
# ============================================================

def update(frame):

    for i, particle in enumerate(particles):

        # ----------------------------------------------------
        # Update position
        # ----------------------------------------------------

        particle.pos.x += particle.vel.x * dt
        particle.pos.y += particle.vel.y * dt


        # ----------------------------------------------------
        # Collision with left/right walls
        # ----------------------------------------------------

        if particle.pos.x >= lx:

            particle.pos.x = lx
            particle.vel.x = -abs(particle.vel.x)

        elif particle.pos.x <= 0:

            particle.pos.x = 0
            particle.vel.x = abs(particle.vel.x)


        # ----------------------------------------------------
        # Collision with bottom/top walls
        # ----------------------------------------------------

        if particle.pos.y >= ly:

            particle.pos.y = ly
            particle.vel.y = -abs(particle.vel.y)

        elif particle.pos.y <= 0:

            particle.pos.y = 0
            particle.vel.y = abs(particle.vel.y)


        # ----------------------------------------------------
        # Save trajectory
        # ----------------------------------------------------

        x_history[i].append(particle.pos.x)
        y_history[i].append(particle.pos.y)


        # ----------------------------------------------------
        # Update particle position on graph
        # ----------------------------------------------------

        particle_plots[i].set_data(
            [particle.pos.x],
            [particle.pos.y]
        )


        # ----------------------------------------------------
        # Update trajectory
        # ----------------------------------------------------

        trajectory_plots[i].set_data(
            x_history[i],
            y_history[i]
        )


    return particle_plots + trajectory_plots


# ============================================================
# Create animation
# ============================================================

ani = anim.FuncAnimation(
    fig,
    update,
    frames=1000,
    interval=20,
    blit=True
)


plt.show()
```
